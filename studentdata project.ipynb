```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv(r"student_data.csv")
df
```

AVERAGE MARKS IN  MATH AND SCIENCE


```python
df["average"] =(df["Maths"]+ df["Science"]+ df["English"]) /3
df.head()


```

TOP STUDENTS

```python
top_student=df.loc[df['average'].idxmax()]
print(top_student)
```

```python
duplicate_name=df.drop_duplicates(subset="Name")
print(duplicate_name)
```

ADDING GRADE

```python
def grade(avg):
    if avg >=90:
        return "A"
    elif avg >=80:
        return "B"
    elif avg >=70:
        return "C"
    else:
        return "D"
Grade=df["average"].apply(grade)
print(Grade)
```
